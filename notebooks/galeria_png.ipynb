# Galeria dos PNGs gerados
Imagens reconstruídas dos CSVs, com estilos definidos em `parametros.json`. As definições de cores e layout estão no gerador; não são reproduções certificadas dos originais. Altere `indice` para ver outros exemplos.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

raiz = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'dados_projeto_multi_modais').is_dir())
import json
config = json.loads((raiz / 'configs' / 'data.json').read_text(encoding='utf-8'))
pasta = raiz / config['images']
indice = 1  # 1 a 120
assert 1 <= indice <= 120
tipos = ['BAR', 'LIN', 'DIS', 'PIZ', 'HIS', 'BOX']
fig, axes = plt.subplots(3, 2, figsize=(18, 17), layout='constrained')
for ax, tipo in zip(axes.flat, tipos):
    caminho = pasta / f'{tipo}-{indice:03d}.png'
    ax.imshow(mpimg.imread(caminho))
    ax.set_title(caminho.name)
    ax.axis('off')
plt.show()
```
